```python
import os
lessons = [name for name in os.listdir() if (name[:6] == 'lesson' and ('12' not in name))]
list(enumerate(lessons))
```

```python
import json 

for i in range(len(lessons)):
    # print()
    file_name = lessons[i]+r'/index.json'
    
    f = open(file_name, 'r', encoding='utf-8')
    text = f.read()

        
    print(i, lessons[i])
    dct = json.loads(text)
    # print(dct["layoutType"])
    dct["layoutType"] = "LAYOUT_TYPE_SIMPLE"

    json_res = json.dumps(dct, indent=4, ensure_ascii=False)

    f = open(file_name, 'w', encoding='utf-8')
    print(json_res)
    f.write(str(json_res))

        # print(json_res)


        # f = open(file_name, 'r', encoding='utf-8')
        # # print(_)
        # # print(f.read())
```
