```python
import numpy as np
import os
import re
import pandas as pd
```

```python
path = os.getcwd()
lessons = [folder for folder in os.listdir(path) if folder.split('_')[0] == 'lesson']
lessons
```

```python
len(lessons)
```

```python
pattern_1 = r'^problem_\d{1,2}\.md$'
pattern_2 = r'lesson_\d+_\d+_'

dct = {}

for lesson in lessons:
    # print(lesson)
    files = [name for name in os.listdir(lesson) if re.match(pattern_1, name)]
    # print(files)
    # print('number of problems', len(files))

    number = list(map(int, lesson.split('_')[1:3]))


    
    title = re.sub(pattern_2, '', lesson)

    # print(number)
    # print(title)
    # print()
    for i, problem in enumerate(files):

        path = os.path.join(lesson, problem)
        # print(path)
        file = open(path, 'r', encoding="utf8").read()
        file_len = len(file)

        dct_ = {'folder':lesson, 'lesson_n':number[0], 'series_n':number[1], 'problem_n':i+1, 'title':title, 'name':problem, 'file_len':file_len}
        dct[tuple(number+[i])] = dct_

df = pd.DataFrame(dct).T.sort_index()
```

```python
df.to_csv('ploblem_len.csv')
```

```python
count = {}

for lesson_n in (list(set(list(df['lesson_n'])))):
    series_n = len(list(set(list(df[df['lesson_n'] == lesson_n]['series_n']))))
    problem_n = len(df[df['lesson_n'] == lesson_n])

    total_len = (df[df['lesson_n'] == lesson_n]['file_len'].sum())
    
    count[lesson_n] = {'lesson #':lesson_n, 'series number':series_n, 'problems number':problem_n, 'total_len':total_len}
```

```python

# print(count)

count_df = pd.DataFrame(count).T
count_df.to_csv('count.csv')
count_df
```
